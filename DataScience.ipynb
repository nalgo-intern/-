## Plot Flavor Sales for Each Age Group

### Subtask:
年代区分（「29歳以下」、「30歳～49歳」、「50歳以上」）ごとに、各アイスクリームの味の販売数を棒グラフで可視化します。これにより、年代ごとの味の好みの違いを比較しやすくします。既に`cross_tab`というクロス集計表があるので、これを利用してプロットを作成します。


**Reasoning**:
The subtask is to visualize the flavor sales for each age group using a bar chart from the `cross_tab` DataFrame. I will follow the instructions to create the plot, including setting the figure size, plotting the bar chart, adding a title, setting axis labels, rotating x-axis labels, adding a legend, ensuring a tight layout, and displaying the plot.



```python
plt.figure(figsize=(12, 7))
cross_tab.plot(kind='bar', ax=plt.gca())
plt.title("年代ごとのアイスクリームの味の販売数")
plt.xlabel("年代")
plt.ylabel("販売数")
plt.xticks(rotation=45, ha='right')
plt.legend(title="味")
plt.tight_layout()
plt.show()
```

```python
df_1_11 = df[df["日付"].dt.month.isin(range(1, 12))]
df_weekdays = df_1_11[df_1_11["曜日"].isin([0, 1, 2, 3, 4])]
sales = df_weekdays.groupby("日付").size()
temps = df_weekdays.groupby("日付")["気温"].mean()
temp_sales = pd.DataFrame({
    "販売数":sales,
    "気温":temps
})
plt.figure(figsize=(10, 6))
sns.regplot(x="気温", y="販売数", data=temp_sales)
plt.title("気温と1日のアイスクリーム販売数")
plt.tight_layout()
plt.show()
```

```python
import statsmodels.api as sm
X = temp_sales["気温"]
y = temp_sales["販売数"]
X = sm.add_constant(X)
model = sm.OLS(y, X).fit()
p_value = model.pvalues["気温"]
coef = model.params["気温"]
print(f"p値:{p_value:.4f}")
print(f"傾き:{coef:.4f}")
alpha = 0.05
if p_value < alpha:
  print("有意差あり：気温が販売数に与える影響は統計的に有意です")
  if coef > 0:
    print("気温が高いほど、販売数が増加する傾向があります")
  else:
    print("気温が高いほど、販売数が減少する傾向があります")
else:
  print("有意差なし：気温が販売数に与える影響は統計的に有意ではありません")
```
